## Resumen del capítulo 3 — Análisis de algoritmos (Goodrich, Tamassia & Goldwasser)

### 3.1 Estudios experimentales

Una primera forma de evaluar un algoritmo es **medir su tiempo de ejecución real** (por ejemplo con `time.time()` o `timeit`), variando el tamaño `n` de la entrada y graficando el resultado.

Esta estrategia tiene limitaciones importantes:

- El tiempo depende del **hardware, del sistema operativo y del lenguaje** en que se ejecuta, por lo que dos mediciones solo son comparables si se hicieron en las mismas condiciones.
- Solo se conoce el comportamiento para los **tamaños de entrada efectivamente probados**; no dice nada sobre entradas mucho más grandes.
- Requiere tener el algoritmo **completamente implementado** antes de poder compararlo con otro.

Por estas razones, se prefiere un análisis **independiente de la máquina**: contar cuántas **operaciones primitivas** (asignaciones, comparaciones, operaciones aritméticas, accesos a listas, llamadas a funciones) ejecuta el algoritmo en función de `n`, y expresar ese conteo con notación asintótica.

### 3.2 Las siete funciones más usadas en el análisis de algoritmos

| Función | Nombre | Ejemplo típico |
|---|---|---|
| $1$ | Constante | Acceder a un elemento de una lista por índice |
| $\log n$ | Logarítmica | Búsqueda binaria |
| $n$ | Lineal | Recorrer una lista una sola vez |
| $n \log n$ | Linealítmica | Mergesort, Heapsort |
| $n^2$ | Cuadrática | Dos ciclos anidados sobre la entrada |
| $n^3$ | Cúbica | Tres ciclos anidados sobre la entrada |
| $2^n$ | Exponencial | Generar todos los subconjuntos de un conjunto de tamaño $n$ |

Para $n$ suficientemente grande, estas funciones quedan ordenadas de menor a mayor crecimiento:

$$1 \;<\; \log n \;<\; n \;<\; n\log n \;<\; n^2 \;<\; n^3 \;<\; 2^n$$

### 3.3 Notación asintótica

**Big-O (cota superior, "no crece más rápido que"):** $f(n)$ es $O(g(n))$ si existen constantes $c > 0$ y $n_0 \ge 0$ tales que

$$f(n) \le c \cdot g(n) \quad \text{para todo } n \ge n_0$$

Es la notación más usada en la práctica porque acota el **peor caso** de crecimiento de un algoritmo.

**Big-Omega (cota inferior, "no crece más lento que"):** $f(n)$ es $\Omega(g(n))$ si existen $c > 0$, $n_0$ tales que $f(n) \ge c \cdot g(n)$ para todo $n \ge n_0$.

**Big-Theta (cota ajustada):** $f(n)$ es $\Theta(g(n))$ si $f(n)$ es simultáneamente $O(g(n))$ y $\Omega(g(n))$; es decir, cuando $g(n)$ describe fielmente el crecimiento de $f(n)$ tanto por arriba como por abajo.

#### Reglas generales para simplificar una cota Big-O

- Si $f(n)$ es un polinomio de grado $d$ (con coeficiente principal positivo), entonces $f(n)$ es $O(n^d)$: se descartan los términos de menor orden y los coeficientes.
- Se usa siempre la función **más pequeña posible** que siga siendo una cota válida (por ejemplo, $2n+3$ es $O(n)$, no hace falta escribir $O(n^2)$).
- No se incluyen constantes multiplicativas ni la base de los logaritmos: $\log_2 n$, $\log_{10} n$, etc. son todos $\Theta(\log n)$.

Estas ideas se aplican en detalle más abajo sobre los ejemplos de `prefix_average`, `find_max`, `disjointness` y `uniqueness`.

## Análisis de complejidad

| Algoritmo | Complejidad | Razón |
|---|---|---|
| `promedios` | O(n²) | `sum(L[0:i+1])` recorre hasta `i` elementos en cada iteración |
| `prefix_average1` | O(n²) | Dos ciclos anidados explícitos |
| `prefix_average2` | O(n²) | `sum(S[0:j+1])` es O(j) en cada iteración del ciclo externo |
| `prefix_average3` | O(n) | Un solo ciclo con acumulador; elimina la suma repetida |

`prefix_average3` es la versión óptima: mantiene una suma acumulada (`total`) que se actualiza en O(1) por iteración.

### Análisis detallado: `prefix_average1`

```python
def prefix_average1(S):
    n = len(S)
    A = [0] * n
    for j in range(n):         # (1)
        total = 0
        for i in range(j + 1): # (2)
            total += S[i]      # (3)
        A[j] = total / (j+1)   # (4)
    return A
```

Solo se cuentan **operaciones aritméticas** y **comparaciones**:

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `j < n` | $n$ |
| (2) | Comparación `i < j+1` | $\displaystyle\sum_{j=0}^{n-1}(j+1) = \frac{n(n+1)}{2}$ |
| (3) | Suma `total += S[i]` | $\displaystyle\sum_{j=0}^{n-1}(j+1) = \frac{n(n+1)}{2}$ |
| (4) | División `total / (j+1)` | $n$ |

**Total de operaciones:**

$$T(n) = n + \frac{n(n+1)}{2} + \frac{n(n+1)}{2} + n = 2n + n(n+1) = n^2 + 3n$$

**Conclusión:** $T(n) = \boxed{O(n^2)}$

### Análisis detallado: `prefix_average3`

```python
def prefix_average3(S):
    n = len(S)
    A = [0] * n
    total = 0
    for j in range(n):       # (1)
        total += S[j]        # (2)
        A[j] = total / (j+1) # (3)
    return A
```

Solo se cuentan **operaciones aritméticas** y **comparaciones**:

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `j < n` | $n$ |
| (2) | Suma `total += S[j]` | $n$ |
| (3) | División `total / (j+1)` | $n$ |

**Total de operaciones:**

$$T(n) = n + n + n = 3n$$

**Conclusión:** $T(n) = \boxed{O(n)}$

A diferencia de `prefix_average1`, aquí el ciclo interno desaparece: el acumulador `total` permite actualizar la suma en O(1) por iteración, pasando de $n^2 + 3n$ operaciones a solo $3n$.

### Análisis mejor y peor caso: `find_max`

```python
def find_max(data):
    biggest = data[0]
    for val in data:      # recorre los n elementos
        if val > biggest: # (1)
            biggest = val # (2)
    return biggest
```

Se cuentan la comparación `val > biggest` y la asignación `biggest = val`.

---

**Mejor caso — lista decreciente** `[n, n−1, …, 2, 1]`

`biggest = data[0]` recibe el valor máximo desde el inicio. Todas las comparaciones dan **False**; la asignación (2) nunca se ejecuta.

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `val > biggest` | $n$ |
| (2) | Asignación `biggest = val` | $0$ |

$$T_{\text{mejor}}(n) = n$$

---

**Peor caso — lista creciente** `[1, 2, …, n−1, n]`

`biggest` comienza en el valor mínimo. La comparación da **True** para los elementos $2, 3, \ldots, n$, por lo que la asignación (2) se ejecuta $n-1$ veces.

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `val > biggest` | $n$ |
| (2) | Asignación `biggest = val` | $n - 1$ |

$$T_{\text{peor}}(n) = n + (n-1) = 2n - 1$$

---

**Conclusión:**

$$T_{\text{mejor}}(n) = n \qquad T_{\text{peor}}(n) = 2n-1$$

$$\boxed{O(n)}\ \text{en ambos casos}$$

Ahora sí existe una diferencia numérica concreta: en el peor caso se realizan casi el doble de operaciones. Sin embargo, ambos son $O(n)$ porque la diferencia es un factor constante.

1. Goodrich 3.2 The number of operations executed by algorithms A and B is $8n\log n$ and
$2n^2$, respectively. Determine n0 such that A is better than B for $n ≥ n_0$.

**Solución (con $\le$)**

Queremos el menor $n_0$ tal que A sea "al menos tan buena" que B (es decir, $8n\log n \le 2n^2$) para todo $n \ge n_0$.

$$8n\log_2 n \le 2n^2$$

Dividiendo ambos lados por $2n$ (válido para $n>0$):

$$4\log_2 n \le n$$

Evaluamos algunos valores:

- $n=16$: $4\log_2 16 = 4\cdot 4 = 16$ → $16 \le 16$ es **verdadero** (se cumple con igualdad).
- $n=15$: $4\log_2 15 \approx 4\cdot 3.907 = 15.63$ → $15.63 \le 15$ es **falso**.

Como la función $n - 4\log_2 n$ es creciente para $n\ge 1$, una vez que la desigualdad se cumple sigue cumpliéndose para todo $n$ mayor.

**Respuesta:** $n_0 = 16$. Para todo $n \ge 16$, se cumple $8n\log n \le 2n^2$ (con igualdad exacta en $n=16$).

*Nota:* si se exige la desigualdad **estricta** ($8n\log n < 2n^2$), el punto de empate en $n=16$ no cuenta y el menor $n_0$ pasa a ser $17$.

````python
# Verificación numérica de n0
import math

def A(n):
    return 8 * n * math.log2(n)

def B(n):
    return 2 * n**2

for n in range(13, 18):
    print(n, A(n), B(n), "A <= B" if A(n) <= B(n) else "A > B")
````

````python
# Gráfica: punto de corte entre 8n*log2(n) y 2n^2
import numpy as np
import matplotlib.pyplot as plt

n = np.linspace(1, 30, 500)
fA = 8 * n * np.log2(n)
fB = 2 * n**2

n0 = 16
yA_n0 = 8 * n0 * math.log2(n0)

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(n, fA, color="#1B6FC9", linewidth=2, label=r"A: $8n\log_2 n$")
ax.plot(n, fB, color="#D9720C", linewidth=2, label=r"B: $2n^2$")

# Punto de corte n0
ax.plot(n0, yA_n0, "o", color="#333333", markersize=8, zorder=5)
ax.axvline(n0, color="#888888", linestyle="--", linewidth=1)
ax.annotate(
    f"$n_0={n0}$",
    xy=(n0, yA_n0),
    xytext=(n0 + 1.5, yA_n0 + 200),
    fontsize=11,
)

ax.set_xlabel("n")
ax.set_ylabel("Número de operaciones")
ax.set_title("Ejercicio 3.3 — Intersección de A y B")
ax.legend(frameon=False)
ax.grid(True, linewidth=0.4, alpha=0.5)
ax.set_xlim(1, 30)
ax.set_ylim(0, max(fB[-1], fA[-1]) * 1.05)

plt.tight_layout()
plt.show()
````

2. 3.3 Goodrich The number of operations executed by algorithms A and B is $40n^2$ and
$2n^3$, respectively. Determine $n_0$ such that A is better than B for $n \geq n_0$.

**Solución**

Este ejercicio es idéntico al 3.2 anterior: $40n^2$ vs $2n^3$.

$$40n^2 \le 2n^3 \;\Longleftrightarrow\; 20 \le n$$

Evaluando:

- $n=19$: $40n^2 = 14440 \le 2n^3$ = 13718 → **falso**.
- $n=20$: $40n^2 = 16000 \le 2n^3$ = 16000 → **Verdadero** (igualdad exacta).

Como $2n^3 - 40n^2$ es creciente para $n \ge 14$, la desigualdad se mantiene para todo $n \ge 20$.

**Respuesta:** $n_0 = 20$ (con $\leq$). Si se pide la desigualdad estricta ($<$), entonces $n_0 = 21$, pues en $n=20$ hay empate.

````python
import numpy as np
import matplotlib.pyplot as plt

n = np.linspace(1, 30, 500)
fA = 40*n**2
fB = 2*n**3

n0 = 20
yA_n0 = 40*n0**2

fig1, ax = plt.subplots(figsize=(7, 5))

ax.plot(n, fA, color="#1B6FC9", linewidth=2, label=r"A: $40n^2$")
ax.plot(n, fB, color="#D9720C", linewidth=2, label=r"B: $2n^3$")

# Punto de corte n0
ax.plot(n0, yA_n0, "o", color="#333333", markersize=8, zorder=5)
ax.axvline(n0, color="#888888", linestyle="--", linewidth=1)
ax.annotate(
    f"$n_0={n0}$",
    xy=(n0, yA_n0),
    xytext=(n0 + 1.5, yA_n0 + 200),
    fontsize=11,
)

ax.set_xlabel("n")
ax.set_ylabel("Número de operaciones")
ax.set_title("Ejercicio 3.3 — Intersección de A y B")
ax.legend(frameon=False)
ax.grid(True, linewidth=0.4, alpha=0.5)
ax.set_xlim(1, 30)
ax.set_ylim(0, max(fB[-1], fA[-1]) * 1.05)

plt.tight_layout()
plt.show()
````

3. Goodrich 3.45 A sequence $S$ contains $n - 1$ unique integers in the range $[0, n-1]$, that is, there is one number from this range that is not in $S$. Design an $O(n)$-time algorithm for finding that number. You are only allowed to use $O(1)$ additional space besides the sequence $S$ itself.

- Escribir un código en python `faltante1(S)` de orden $O(n^2)$ que resuelva el problema.
- Escribir un código en python `faltante(S)` de orden $O(n)$ que resuelva el problema.

````python
def faltante1(S):
    for i in range(0, len(S)+1):
        faltante = True
        for j in S:
            if i == j:
                faltante = False #(el valor i se encuentra en S)
        if faltante == True:
            return i

def faltante(S):
    num = 0
    for i in S:
        num += i
    return len(S)*(len(S) + 1)/2 - num
            
````

````python
# Un ejemplo para verificar
S = [2,3,0] # Falta el número 1
faltante1(S), faltante(S)
````

````python
import random

# Opción 1: elige al azar el número faltante y lo excluye por comprensión
def generar_S(n):
    """Genera una lista de n-1 enteros únicos en [0, n-1], con un valor faltante."""
    faltante = random.randint(0, n - 1)
    S = [i for i in range(n) if i != faltante]
    random.shuffle(S)  # opcional: para que no queden ordenados
    return S, faltante

# Opción 2: no guarda el valor faltante en ninguna variable
def generar_S2(n):
    S = list(range(n))
    S.pop(random.randrange(n))  # elimina un elemento al azar, sin guardar su valor
    return S
````
