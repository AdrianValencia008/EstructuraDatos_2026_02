## Conceptos teóricos


````python
# Vamoa a escribir una función que retorne la lista de promedios sucesivos de una lista
def promedios(L):
    '''Se retorna el promedio hasta la entrada i de una lista L de números'''
    resultado = []
    for i in range(len(L)):
        resultado.append(sum(L[0:i+1])/(i+1))

    return resultado
````

## Análisis de complejidad

| Algoritmo | Complejidad | Razón |
|---|---|---|
| `promedios` | O(n²) | `sum(L[0:i+1])` recorre hasta `i` elementos en cada iteración |
| `prefix_average1` | O(n²) | Dos ciclos anidados explícitos |
| `prefix_average2` | O(n²) | `sum(S[0:j+1])` es O(j) en cada iteración del ciclo externo |
| `prefix_average3` | O(n) | Un solo ciclo con acumulador; elimina la suma repetida |

`prefix_average3` es la versión óptima: mantiene una suma acumulada (`total`) que se actualiza en O(1) por iteración.

### Análisis detallado: `prefix_average1`

```python
def prefix_average1(S):
    n = len(S)
    A = [0] * n
    for j in range(n):         # (1)
        total = 0
        for i in range(j + 1): # (2)
            total += S[i]      # (3)
        A[j] = total / (j+1)   # (4)
    return A
```

Solo se cuentan **operaciones aritméticas** y **comparaciones**:

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `j < n` | $n$ |
| (2) | Comparación `i < j+1` | $\displaystyle\sum_{j=0}^{n-1}(j+1) = \frac{n(n+1)}{2}$ |
| (3) | Suma `total += S[i]` | $\displaystyle\sum_{j=0}^{n-1}(j+1) = \frac{n(n+1)}{2}$ |
| (4) | División `total / (j+1)` | $n$ |

**Total de operaciones:**

$$T(n) = n + \frac{n(n+1)}{2} + \frac{n(n+1)}{2} + n = 2n + n(n+1) = n^2 + 3n$$

**Conclusión:** $T(n) = \boxed{O(n^2)}$

### Análisis detallado: `prefix_average3`

```python
def prefix_average3(S):
    n = len(S)
    A = [0] * n
    total = 0
    for j in range(n):       # (1)
        total += S[j]        # (2)
        A[j] = total / (j+1) # (3)
    return A
```

Solo se cuentan **operaciones aritméticas** y **comparaciones**:

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `j < n` | $n$ |
| (2) | Suma `total += S[j]` | $n$ |
| (3) | División `total / (j+1)` | $n$ |

**Total de operaciones:**

$$T(n) = n + n + n = 3n$$

**Conclusión:** $T(n) = \boxed{O(n)}$

A diferencia de `prefix_average1`, aquí el ciclo interno desaparece: el acumulador `total` permite actualizar la suma en O(1) por iteración, pasando de $n^2 + 3n$ operaciones a solo $3n$.

### Análisis mejor y peor caso: `find_max`

```python
def find_max(data):
    biggest = data[0]
    for val in data:      # recorre los n elementos
        if val > biggest: # (1)
            biggest = val # (2)
    return biggest
```

Se cuentan la comparación `val > biggest` y la asignación `biggest = val`.

---

**Mejor caso — lista decreciente** `[n, n−1, …, 2, 1]`

`biggest = data[0]` recibe el valor máximo desde el inicio. Todas las comparaciones dan **False**; la asignación (2) nunca se ejecuta.

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `val > biggest` | $n$ |
| (2) | Asignación `biggest = val` | $0$ |

$$T_{\text{mejor}}(n) = n$$

---

**Peor caso — lista creciente** `[1, 2, …, n−1, n]`

`biggest` comienza en el valor mínimo. La comparación da **True** para los elementos $2, 3, \ldots, n$, por lo que la asignación (2) se ejecuta $n-1$ veces.

| Línea | Operación | Ejecuciones |
|:---:|---|---|
| (1) | Comparación `val > biggest` | $n$ |
| (2) | Asignación `biggest = val` | $n - 1$ |

$$T_{\text{peor}}(n) = n + (n-1) = 2n - 1$$

---

**Conclusión:**

$$T_{\text{mejor}}(n) = n \qquad T_{\text{peor}}(n) = 2n-1$$

$$\boxed{O(n)}\ \text{en ambos casos}$$

Ahora sí existe una diferencia numérica concreta: en el peor caso se realizan casi el doble de operaciones. Sin embargo, ambos son $O(n)$ porque la diferencia es un factor constante.

# 10 de junio: three-way set disjointness

````python
def disjointness(L, L1, L2):
    '''Retorna True si las tres listas no tienen repeticiones'''
    for i in L:
        for j in L1:
            for k in L2:
                if i == j == k:
                    return False
    
    return True
````

````python
# The uniqueness problem
def uniqueness(L):
    '''Retorna True si la lista no tiene repeticiones'''
    for i in range(len(L)):
        for j in range(len(L)):
            if i != j and L[i] == L[j]:
                return False

    return True
````

````python
def uniqueness1(L):
    '''Retorna True si la lista no tiene repeticiones'''
    for i in range(len(L)):
        for j in range(i+1, len(L)):
            if L[i] == L[j]:
                return False

    return True
````
