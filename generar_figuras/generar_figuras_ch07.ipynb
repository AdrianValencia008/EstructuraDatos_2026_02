# Generador de figuras — CH07 Listas Enlazadas

Ejecuta cada celda para regenerar las imágenes en `notebooks/assets/`.

> Este archivo está en `.gitignore` y no se versiona.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from pathlib import Path

IMG_DIR = Path('notebooks/assets')
IMG_DIR.mkdir(parents=True, exist_ok=True)

def sl_node(ax, x, y, val, w=1.8, h=0.75,
            dc='#90CAF9', pc='#E3F2FD', ec='#0D47A1'):
    dw, pw = w*0.60, w*0.40
    ax.add_patch(patches.Rectangle((x,y),    dw, h, fc=dc, ec=ec, lw=1.8, zorder=2))
    ax.add_patch(patches.Rectangle((x+dw,y), pw, h, fc=pc, ec=ec, lw=1.8, zorder=2))
    ax.text(x+dw/2, y+h/2, str(val), ha='center', va='center',
            fontsize=13, fontweight='bold', color=ec, zorder=3)
    cx, cy = x+dw+pw/2, y+h/2
    ax.plot(cx, cy, 'o', color=ec, markersize=6, zorder=4)
    return cx, cy
```

```python
# Figura 1 — Lista Enlazada Simple con head y tail
fig, ax = plt.subplots(figsize=(13, 3.8))
ax.axis('off'); ax.set_xlim(-2.2, 12.0); ax.set_ylim(-0.5, 2.8)
ax.set_title('Lista Enlazada Simple (Singly Linked List)',
             fontsize=14, fontweight='bold', color='#1A237E', pad=8)

nodes = ['A', 'B', 'C', 'D']
w, h, gap, y0 = 1.8, 0.75, 0.9, 0.65
xs = [i*(w+gap) for i in range(4)]

# head — flecha horizontal desde la izquierda
ax.annotate('', xy=(xs[0], y0+h/2), xytext=(-1.8, y0+h/2),
            arrowprops=dict(arrowstyle='->', color='#2E7D32', lw=2.2))
ax.text(-1.9, y0+h+0.1, 'head', fontsize=10, color='#2E7D32', fontweight='bold', ha='center')

# tail — flecha vertical desde arriba apuntando al último nodo
tx = xs[-1] + w*0.30   # centro de la parte "elemento" del último nodo
ax.annotate('', xy=(tx, y0+h+0.02), xytext=(tx, y0+h+0.85),
            arrowprops=dict(arrowstyle='->', color='#6A1B9A', lw=2.2))
ax.text(tx, y0+h+1.02, 'tail', fontsize=10, color='#6A1B9A', fontweight='bold', ha='center')

for i, (val, x) in enumerate(zip(nodes, xs)):
    cx, cy = sl_node(ax, x, y0, val, w=w, h=h)
    if i < len(nodes)-1:
        ax.annotate('', xy=(xs[i+1], cy), xytext=(cx, cy),
                    arrowprops=dict(arrowstyle='->', color='#1565C0', lw=1.8))
    else:
        ax.annotate('', xy=(x+w+0.3, cy), xytext=(cx, cy),
                    arrowprops=dict(arrowstyle='->', color='#1565C0', lw=1.8))
        ax.text(x+w+0.35, cy, 'None', fontsize=11, va='center',
                color='#B71C1C', fontweight='bold')

ax.text(xs[0]+w*0.30, y0-0.22, 'elemento', ha='center', fontsize=8, color='#78909C', style='italic')
ax.text(xs[0]+w*0.80, y0-0.22, 'next',     ha='center', fontsize=8, color='#78909C', style='italic')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch07_singly_linked.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch07_singly_linked.png')
```

```python
# Figura 2 — Lista Enlazada Circular
fig, ax = plt.subplots(figsize=(11, 4.2))
ax.axis('off'); ax.set_xlim(-1.5, 10.0); ax.set_ylim(-1.6, 2.8)
ax.set_title('Lista Enlazada Circular (Circularly Linked List)',
             fontsize=14, fontweight='bold', color='#1A237E', pad=8)

cnodes  = ['A', 'B', 'C']
ccolors = ['#FFCC80', '#FFE082', '#FF8F00']
cw, ch_, cgap, cy0 = 1.8, 0.75, 0.9, 0.8
cxs = [i*(cw+cgap) for i in range(3)]

tx = cxs[-1] + cw*0.30
ax.annotate('', xy=(tx, cy0+ch_+0.02), xytext=(tx, cy0+ch_+0.95),
            arrowprops=dict(arrowstyle='->', color='#6A1B9A', lw=2.2))
ax.text(tx, cy0+ch_+1.12, 'tail', fontsize=10, color='#6A1B9A', fontweight='bold', ha='center')

for i, (val, x, clr) in enumerate(zip(cnodes, cxs, ccolors)):
    cx, cy = sl_node(ax, x, cy0, val, w=cw, h=ch_, dc=clr, pc='#FFF9C4', ec='#E65100')
    if i < len(cnodes)-1:
        ax.annotate('', xy=(cxs[i+1], cy), xytext=(cx, cy),
                    arrowprops=dict(arrowstyle='->', color='#E65100', lw=1.8))

last_cx = cxs[-1] + cw*0.60 + cw*0.40/2
ax.annotate('', xy=(cxs[0], cy0+ch_/2), xytext=(last_cx, cy0+ch_/2),
            arrowprops=dict(arrowstyle='->', color='#E65100', lw=2.0,
                           connectionstyle='arc3,rad=-0.55'))
mid_x = (cxs[0] + cxs[-1]+cw) / 2
ax.text(mid_x, cy0-1.25, 'tail._next  apunta al frente de la cola',
        ha='center', fontsize=9, color='#E65100', style='italic')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch07_circularly_linked.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch07_circularly_linked.png')
```

```python
# Figura 3 — Lista Doblemente Enlazada con Centinelas
fig, ax = plt.subplots(figsize=(15, 4.0))
ax.axis('off'); ax.set_xlim(-0.4, 14.8); ax.set_ylim(-0.4, 2.4)
ax.set_title('Lista Doblemente Enlazada con Centinelas',
             fontsize=14, fontweight='bold', color='#1A237E', pad=8)

all_nodes = [('HEADER', True), ('A', False), ('B', False), ('C', False), ('TRAILER', True)]
nw, nh, ngap, ny0 = 2.2, 0.75, 0.8, 0.7
pw_c = nw*0.22; ew_c = nw*0.56
nxs  = [i*(nw+ngap) for i in range(5)]
prev_right = None

for i, (val, is_sent) in enumerate(all_nodes):
    x = nxs[i]; ec_ = '#2E7D32'
    fc_d = '#F5F5F5' if is_sent else '#A5D6A7'
    fc_p = '#EEEEEE' if is_sent else '#E8F5E9'
    ax.add_patch(patches.Rectangle((x, ny0),           pw_c, nh, fc=fc_p, ec=ec_, lw=1.8, zorder=2))
    ax.add_patch(patches.Rectangle((x+pw_c, ny0),      ew_c, nh, fc=fc_d, ec=ec_, lw=1.8, zorder=2))
    ax.add_patch(patches.Rectangle((x+pw_c+ew_c, ny0), pw_c, nh, fc=fc_p, ec=ec_, lw=1.8, zorder=2))
    ax.text(x+pw_c+ew_c/2, ny0+nh/2, val, ha='center', va='center',
            fontsize=8 if is_sent else 13, fontweight='bold', color=ec_, zorder=3)
    lc = x+pw_c/2; rc = x+pw_c+ew_c+pw_c/2
    if i == 0:
        ax.text(lc, ny0+nh/2, '∅', ha='center', va='center',
                fontsize=12, color='#B71C1C', fontweight='bold', zorder=4)
    else:
        ax.plot(lc, ny0+nh/2, 'o', color=ec_, markersize=5, zorder=4)
    if i == len(all_nodes)-1:
        ax.text(rc, ny0+nh/2, '∅', ha='center', va='center',
                fontsize=12, color='#B71C1C', fontweight='bold', zorder=4)
    else:
        ax.plot(rc, ny0+nh/2, 'o', color=ec_, markersize=5, zorder=4)
    if prev_right is not None:
        yf = ny0+nh*0.70; yb = ny0+nh*0.30
        ax.annotate('', xy=(x, yf), xytext=(prev_right, yf),
                    arrowprops=dict(arrowstyle='->', color=ec_, lw=1.5))
        ax.annotate('', xy=(prev_right, yb), xytext=(x, yb),
                    arrowprops=dict(arrowstyle='->', color=ec_, lw=1.5))
    prev_right = x+nw

ri = 1
ax.text(nxs[ri]+pw_c/2,           ny0-0.22, 'prev', ha='center', fontsize=8, color='#78909C', style='italic')
ax.text(nxs[ri]+pw_c+ew_c/2,      ny0-0.22, 'elem', ha='center', fontsize=8, color='#78909C', style='italic')
ax.text(nxs[ri]+pw_c+ew_c+pw_c/2, ny0-0.22, 'next', ha='center', fontsize=8, color='#78909C', style='italic')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch07_doubly_linked.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch07_doubly_linked.png')
```
