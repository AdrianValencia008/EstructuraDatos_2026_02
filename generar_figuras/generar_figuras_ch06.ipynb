# Generador de figuras — CH06 Pilas, Colas y Deques

Ejecuta este notebook para regenerar las imágenes estáticas que usa `ch06_teoria.ipynb`.
Las imágenes se guardan en `notebooks/assets/`.

> **Nota:** este archivo está en `.gitignore` y no se versiona.

```python
import matplotlib
matplotlib.use('Agg')   # sin ventana de display
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
from pathlib import Path

IMG_DIR = Path('notebooks/assets')
IMG_DIR.mkdir(parents=True, exist_ok=True)
print('Directorio:', IMG_DIR.resolve())
```

```python
# Figura 1 — Stack vs Queue
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5.5))
fig.suptitle('Comparación: Pila vs Cola', fontsize=15, fontweight='bold', y=1.01)

ax1.set_xlim(0, 5); ax1.set_ylim(-0.8, 8.0); ax1.axis('off')
ax1.set_title('Stack (Pila) — LIFO', fontsize=13, fontweight='bold', color='#1A237E')

elems   = ['A', 'B', 'C', 'D']
bcolors = ['#BBDEFB', '#90CAF9', '#64B5F6', '#42A5F5']
x0, y0, w, h, gap = 1.0, 0.3, 3.0, 1.1, 0.15

for i, (e, c) in enumerate(zip(elems, bcolors)):
    rect = patches.FancyBboxPatch((x0, y0 + i*(h+gap)), w, h,
                                  boxstyle='round,pad=0.07', fc=c, ec='#0D47A1', lw=2)
    ax1.add_patch(rect)
    ax1.text(x0+w/2, y0+i*(h+gap)+h/2, e,
             ha='center', va='center', fontsize=20, fontweight='bold', color='#0D47A1')

top_y = y0 + len(elems)*(h+gap)
mid_x = x0 + w/2

ax1.annotate('', xy=(mid_x+0.4, top_y+0.1), xytext=(mid_x+0.4, top_y+1.3),
             arrowprops=dict(arrowstyle='->', color='#2E7D32', lw=2.5))
ax1.text(mid_x+0.6, top_y+0.7, 'push()', fontsize=10, va='center', color='#2E7D32', fontweight='bold')

ax1.annotate('', xy=(mid_x-0.4, top_y+1.3), xytext=(mid_x-0.4, top_y+0.1),
             arrowprops=dict(arrowstyle='->', color='#B71C1C', lw=2.5))
ax1.text(mid_x-1.5, top_y+0.7, 'pop()', fontsize=10, va='center', color='#B71C1C', fontweight='bold')

ax1.text(x0-0.05, top_y-0.05, '← TOP',    fontsize=9.5, color='#B71C1C', va='center', fontweight='bold')
ax1.text(x0-0.05, y0+h/2,     '← BOTTOM', fontsize=9.5, color='#455A64', va='center')
ax1.text(mid_x, -0.5, 'Solo opera por la CIMA',
         ha='center', fontsize=9, style='italic', color='#546E7A')

ax2.set_xlim(-2.0, 10.5); ax2.set_ylim(-1.2, 3.8); ax2.axis('off')
ax2.set_title('Queue (Cola) — FIFO', fontsize=13, fontweight='bold', color='#1A237E')

qelems  = ['A', 'B', 'C', 'D']
qcolors = ['#EF9A9A', '#FFCC80', '#A5D6A7', '#BBDEFB']
qx, qy, qw, qh, qgap = 0.0, 0.8, 1.8, 1.4, 0.2

for i, (e, c) in enumerate(zip(qelems, qcolors)):
    rect = patches.FancyBboxPatch((qx+i*(qw+qgap), qy), qw, qh,
                                  boxstyle='round,pad=0.07', fc=c, ec='#37474F', lw=2)
    ax2.add_patch(rect)
    ax2.text(qx+i*(qw+qgap)+qw/2, qy+qh/2, e,
             ha='center', va='center', fontsize=20, fontweight='bold', color='#212121')

front_x = qx
back_x  = qx + len(qelems)*(qw+qgap) - qgap
mid_y   = qy + qh/2

ax2.annotate('', xy=(-1.6, mid_y), xytext=(front_x, mid_y),
             arrowprops=dict(arrowstyle='->', color='#B71C1C', lw=2.5))
ax2.text(-1.7, mid_y+0.35, 'dequeue()', ha='center', fontsize=10, color='#B71C1C', fontweight='bold')

ax2.annotate('', xy=(back_x, mid_y), xytext=(back_x+1.6, mid_y),
             arrowprops=dict(arrowstyle='->', color='#2E7D32', lw=2.5))
ax2.text(back_x+0.8, mid_y+0.35, 'enqueue()', ha='center', fontsize=10, color='#2E7D32', fontweight='bold')

ax2.text(front_x+qw/2, qy-0.45, 'FRENTE\n(front)', ha='center', fontsize=9, color='#B71C1C', fontweight='bold')
ax2.text(back_x-qw/2,  qy-0.45, 'FINAL\n(back)',   ha='center', fontsize=9, color='#2E7D32', fontweight='bold')
ax2.text((front_x+back_x)/2, -0.9, 'Entra por atrás, sale por el frente',
         ha='center', fontsize=9, style='italic', color='#546E7A')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch06_stack_vs_queue.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch06_stack_vs_queue.png')
```

```python
# Figura 2 — ArrayQueue arreglo circular
fig, ax = plt.subplots(figsize=(7, 7))
ax.set_aspect('equal'); ax.axis('off')
ax.set_title('ArrayQueue — Arreglo circular (capacidad = 8)', fontsize=13, fontweight='bold')

N = 8; R_out, R_in = 2.6, 1.7
data = [None, None, 'A', 'B', 'C', None, None, None]
_front, _size = 2, 3

def mid_angle(i):
    return np.pi/2 - 2*np.pi*i/N - np.pi/N

for i in range(N):
    theta1 = np.degrees(np.pi/2 - 2*np.pi*(i+1)/N)
    theta2 = np.degrees(np.pi/2 - 2*np.pi*i/N)
    is_elem = data[i] is not None
    wedge = patches.Wedge((0,0), R_out, theta1, theta2, width=R_out-R_in,
                          fc='#90CAF9' if is_elem else '#F5F5F5',
                          ec='#1565C0' if is_elem else '#BDBDBD', lw=2)
    ax.add_patch(wedge)
    a  = mid_angle(i)
    mr = (R_out + R_in) / 2
    ax.text(mr*np.cos(a), mr*np.sin(a), data[i] if data[i] else '—',
            ha='center', va='center', fontsize=14, fontweight='bold',
            color='#0D47A1' if is_elem else '#9E9E9E')
    ox, oy = (R_out+0.32)*np.cos(a), (R_out+0.32)*np.sin(a)
    ax.text(ox, oy, f'[{i}]', ha='center', va='center', fontsize=9, color='#546E7A')

fa = mid_angle(_front)
ax.annotate('_front\n(dequeue aquí)',
            xy=((R_out+0.05)*np.cos(fa), (R_out+0.05)*np.sin(fa)),
            xytext=((R_out+1.1)*np.cos(fa), (R_out+1.1)*np.sin(fa)),
            arrowprops=dict(arrowstyle='->', color='#B71C1C', lw=2),
            fontsize=9.5, color='#B71C1C', fontweight='bold', ha='center', va='center')

back_idx = (_front + _size) % N
ba = mid_angle(back_idx)
ax.annotate('avail\n(enqueue aquí)',
            xy=((R_out+0.05)*np.cos(ba), (R_out+0.05)*np.sin(ba)),
            xytext=((R_out+1.1)*np.cos(ba), (R_out+1.1)*np.sin(ba)),
            arrowprops=dict(arrowstyle='->', color='#2E7D32', lw=2),
            fontsize=9.5, color='#2E7D32', fontweight='bold', ha='center', va='center')

ax.text(0, 0,
        f'_front = {_front}\n_size  = {_size}\navail  = ({_front}+{_size}) % {N} = {back_idx}',
        ha='center', va='center', fontsize=10, color='#37474F', linespacing=1.6,
        bbox=dict(boxstyle='round,pad=0.5', fc='#FFFDE7', ec='#F9A825', lw=1.5))

ax.set_xlim(-5, 5); ax.set_ylim(-5, 5)
plt.tight_layout()
plt.savefig(IMG_DIR / 'ch06_array_queue_circular.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch06_array_queue_circular.png')
```
