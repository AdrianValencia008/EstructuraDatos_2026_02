```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path

IMG_DIR = Path('notebooks/assets')
IMG_DIR.mkdir(parents=True, exist_ok=True)

def circle_node(ax, x, y, lbl, r=0.42, fc='#90CAF9', ec='#0D47A1', fs=12):
    ax.add_patch(plt.Circle((x, y), r, fc=fc, ec=ec, lw=2.0, zorder=3))
    ax.text(x, y, lbl, ha='center', va='center', fontsize=fs,
            fontweight='bold', color=ec, zorder=4)

def edge(ax, x1, y1, x2, y2, r=0.42, color='#1565C0', lw=1.8):
    dx, dy = x2-x1, y2-y1
    d = (dx**2 + dy**2)**0.5
    ax.plot([x1+dx/d*r, x2-dx/d*r], [y1+dy/d*r, y2-dy/d*r],
            color=color, lw=lw, zorder=2)

print('Utilidades cargadas.')
```

```python
# Figura 1 — Árbol General con terminología
fig, ax = plt.subplots(figsize=(14, 7))
ax.axis('off'); ax.set_xlim(-1.2, 13.5); ax.set_ylim(-0.8, 7.5)
ax.set_title('Árbol General — Terminología', fontsize=15, fontweight='bold',
             color='#1A237E', pad=10)

nodes = {
    'E':  (6.5, 6.5, 'Electronics', '#FFCC80', '#E65100'),
    'Co': (2.5, 4.8, 'Computers',   '#A5D6A7', '#2E7D32'),
    'Ph': (6.5, 4.8, 'Phones',      '#A5D6A7', '#2E7D32'),
    'TV': (10.5,4.8, 'TV',          '#CE93D8', '#6A1B9A'),
    'La': (1.2, 3.0, 'Laptop',      '#CE93D8', '#6A1B9A'),
    'De': (3.8, 3.0, 'Desktop',     '#CE93D8', '#6A1B9A'),
    'Mo': (5.2, 3.0, 'Mobile',      '#CE93D8', '#6A1B9A'),
    'Vo': (7.8, 3.0, 'VoIP',        '#CE93D8', '#6A1B9A'),
}
r = 0.55
edges_ = [('E','Co'),('E','Ph'),('E','TV'),('Co','La'),('Co','De'),('Ph','Mo'),('Ph','Vo')]

for a1,a2 in edges_:
    x1,y1=nodes[a1][0],nodes[a1][1]; x2,y2=nodes[a2][0],nodes[a2][1]
    edge(ax,x1,y1,x2,y2,r=r,color='#455A64',lw=2.2)
for k,(x,y,lbl,fc,ec) in nodes.items():
    circle_node(ax,x,y,lbl,r=r,fc=fc,ec=ec,fs=9)

ax.annotate('raíz (profundidad 0)', xy=(6.5,6.5+r), xytext=(9.2,7.2),
            fontsize=9, color='#E65100', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#E65100', lw=1.4))
ax.annotate('nodo interno (profundidad 1)', xy=(2.5,4.8+r), xytext=(-0.8,5.8),
            fontsize=9, color='#2E7D32', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#2E7D32', lw=1.4))
ax.annotate('hoja (profundidad 2)', xy=(1.2,3.0-r), xytext=(-0.8,1.8),
            fontsize=9, color='#6A1B9A', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#6A1B9A', lw=1.4))
ax.annotate('hoja (profundidad 1)', xy=(10.5,4.8-r), xytext=(11.5,3.5),
            fontsize=9, color='#6A1B9A', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#6A1B9A', lw=1.4))
for depth, y, label in [(0,6.5,'prof. 0'),(1,4.8,'prof. 1'),(2,3.0,'prof. 2')]:
    ax.plot([12.6,12.6],[y-0.2,y+0.2],color='#607D8B',lw=1.5)
    ax.text(12.8, y, label, fontsize=8, va='center', color='#607D8B', style='italic')
ax.text(6.5,-0.4,'altura del árbol = 2  |  padre(Laptop) = Computers  |  hermanos(Laptop, Desktop)',
        ha='center',fontsize=9,color='#37474F',style='italic')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch08_general_tree.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch08_general_tree.png')
```

```python
# Figura 2 — Árbol Binario con etiquetas izq/der
fig, ax = plt.subplots(figsize=(11, 6.5))
ax.axis('off'); ax.set_xlim(-0.5, 10.5); ax.set_ylim(-0.8, 6.2)
ax.set_title('Árbol Binario — estructura y terminología', fontsize=14,
             fontweight='bold', color='#1A237E', pad=10)

pos = {'A':(5.0,5.2),'B':(2.5,3.5),'C':(7.5,3.5),
       'D':(1.2,1.8),'E':(3.8,1.8),'F':(6.2,1.8),'G':(8.8,1.8)}
edgs = [('A','B'),('A','C'),('B','D'),('B','E'),('C','F'),('C','G')]
r = 0.42
ROOT_C='#FFCC80'; ROOT_E='#E65100'
INT_C='#A5D6A7';  INT_E='#2E7D32'
LEAF_C='#CE93D8'; LEAF_E='#6A1B9A'
colors={'A':(ROOT_C,ROOT_E),'B':(INT_C,INT_E),'C':(INT_C,INT_E),
        'D':(LEAF_C,LEAF_E),'E':(LEAF_C,LEAF_E),'F':(LEAF_C,LEAF_E),'G':(LEAF_C,LEAF_E)}

for a1,a2 in edgs:
    edge(ax,*pos[a1],*pos[a2],r=r,color='#455A64')
for lbl,(x,y) in pos.items():
    fc,ec=colors[lbl]; circle_node(ax,x,y,lbl,r=r,fc=fc,ec=ec)

ax.text(3.1,4.6,'hijo\nizq.',ha='center',fontsize=8,color='#1565C0',style='italic')
ax.text(7.0,4.6,'hijo\nder.',ha='center',fontsize=8,color='#1565C0',style='italic')

legend_items=[mpatches.Patch(fc=ROOT_C,ec=ROOT_E,label='Raíz'),
              mpatches.Patch(fc=INT_C, ec=INT_E, label='Nodo interno'),
              mpatches.Patch(fc=LEAF_C,ec=LEAF_E,label='Hoja')]
ax.legend(handles=legend_items,loc='lower right',fontsize=9,framealpha=0.9)
ax.annotate('raíz (sin padre)',xy=(5.0,5.2+r),xytext=(7.5,6.0),fontsize=9,color=ROOT_E,fontweight='bold',
            arrowprops=dict(arrowstyle='->',color=ROOT_E))
ax.text(5.0,-0.5,'Árbol binario completo  |  altura = 2  |  7 nodos  |  4 hojas',
        ha='center',fontsize=9,color='#37474F',style='italic')

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch08_binary_tree.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch08_binary_tree.png')
```

```python
# Figura 3 — Recorridos (2×2)
pos_t = {'A':(3.0,3.2),'B':(1.5,2.0),'C':(4.5,2.0),'D':(0.7,0.7),'E':(2.3,0.7),'F':(5.3,0.7)}
edgs_t = [('A','B'),('A','C'),('B','D'),('B','E'),('C','F')]
traversals = {
    'Preorden\n(Raíz → Izq → Der)': {'A':1,'B':2,'D':3,'E':4,'C':5,'F':6},
    'Inorden\n(Izq → Raíz → Der)':  {'D':1,'B':2,'E':3,'A':4,'C':5,'F':6},
    'Postorden\n(Izq → Der → Raíz)':{'D':1,'E':2,'B':3,'F':4,'C':5,'A':6},
    'BFS\n(por niveles)':            {'A':1,'B':2,'C':3,'D':4,'E':5,'F':6},
}
colors_by_order = ['#EF9A9A','#FFCC80','#FFF59D','#A5D6A7','#90CAF9','#CE93D8']

fig, axes = plt.subplots(2, 2, figsize=(13, 9))
fig.suptitle('Recorridos de Árboles Binarios', fontsize=15, fontweight='bold',
             color='#1A237E', y=1.00)
for ax, (title, order) in zip(axes.flat, traversals.items()):
    ax.axis('off'); ax.set_xlim(0.0, 6.1); ax.set_ylim(0.0, 4.0)
    ax.set_title(title, fontsize=11, fontweight='bold', color='#1A237E', pad=6)
    for a1,a2 in edgs_t:
        edge(ax,*pos_t[a1],*pos_t[a2],r=0.38)
    for lbl,(x,y) in pos_t.items():
        n = order[lbl]
        circle_node(ax,x,y,lbl,r=0.38,fc=colors_by_order[n-1],ec='#37474F')
        ax.text(x+0.30,y+0.30,str(n),fontsize=9,color='#B71C1C',fontweight='bold',zorder=5)

plt.tight_layout()
plt.savefig(IMG_DIR / 'ch08_traversals.png', dpi=130, bbox_inches='tight')
plt.show()
print('Guardado: ch08_traversals.png')
```
