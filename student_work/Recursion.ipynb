# `Clase 3 de agosoto`

1. Crear una función `potencia_dos(n)` que compute $2^n$
2. Una función recursiva que haga lo anterior
3. Una función recursiva pero rapida `potencia_dos_recursiva(n)` que resuelva el problmea anterior

### 1. `potencia_dos(n)`

```python
def potencia_dos(n):
  '''Retorna 2^n'''
  result = 1
  for i in range(n):
    result *= 2
  return result
```

```python
potencia_dos(2)
```

### 2. `potencia_recursiva(n)`

```python
def potencia_recursiva(n):
  if n == 0:
    return 1
  else:
    return 2*potencia_recursiva(n-1)
```

```python
potencia_recursiva(0)
```

### 3. `potencia_dos_recursiva(n)`

```python
def potencia_dos_recursiva(n):
  if n == 0:
    return 1
  else:
    potencia = potencia_dos_recursiva(n//2)
    if n%2 == 0:
      return potencia*potencia
    else:
      return 2*potencia*potencia
```

```python
potencia_dos_recursiva(2)
```

## `Ejercicios`

Escribir una funcion recursiva para encontrar el máximo de una secuencia S

```python
def find_max(S):
  if len(S) == 1:
    return S[0]
  maximo = find_max(S[:-1])
  if maximo > S[-1]:
    return maximo
  else:
    return S[-1]
```

```python
find_max([1,23,4,56,4,32,9,19])
```
